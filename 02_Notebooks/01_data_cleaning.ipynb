## Data-quality checks
Look for inconsistent categories, invalid numerical values and leakage-risk columns.

```python
# ============================================================
# Data Quality Checks
# ============================================================

categorical_columns = [
    "uk_region",
    "support_channel",
    "product_area",
    "priority_initial"
]

for col in categorical_columns:
    print(f"\n{'='*50}")
    print(f"Column: {col}")
    print(f"{'='*50}")

    display(
        df[col]
        .value_counts(dropna=False)
        .head(20)
    )

# Check invalid queue values
negative_queue_count = (
    df["agent_queue_length_at_submit"] < 0
).sum()

print("\nNegative queue values:", negative_queue_count)

# Identify leakage-risk columns
leakage_columns = [
    "resolved_at",
    "escalated_prior_to_resolution",
    "resolution_hours",
    "final_priority"
]

print("\nLeakage-risk columns:")
for col in leakage_columns:
    print(f" - {col}")
```

## Data Quality Assessment Summary

The data-quality review identified several issues that require remediation before modelling. Multiple categorical variables contain inconsistent labels caused by differences in capitalization and formatting. For example, the `support_channel` field contains values such as "Email", "email", and "E-mail", while the `uk_region` and `product_area` fields contain similar inconsistencies. These variations represent the same underlying categories and should be standardized to improve data quality and reduce unnecessary feature fragmentation.

The analysis also identified 64 records with negative values in `agent_queue_length_at_submit`. Since queue length cannot be negative in a real support environment, these values are considered invalid and will be corrected during cleaning.

Missing values were detected in several columns, including `avg_sentiment_score`, `customer_tenure_months`, `product_area`, `priority_initial`, and `submitted_at`. Although the percentage of missing values is relatively low, appropriate treatment is required to ensure data completeness and consistency.

Finally, four variables were identified as potential sources of target leakage: `resolved_at`, `escalated_prior_to_resolution`, `resolution_hours`, and `final_priority`. These features contain information that becomes available during or after ticket resolution and therefore must be excluded from predictive modelling to maintain realistic model performance and avoid leakage.


## Cleaning Decisions

The dataset was cleaned by standardizing categorical values, removing duplicate records, correcting invalid queue length values, converting date columns to datetime format, and handling missing values using simple and reproducible imputation methods. These decisions were made to improve data quality while minimizing information loss.

```python
# ============================================================
# Data Cleaning
# ============================================================

clean = df.copy()

# ------------------------------------------------------------
# Standardise text categories
# ------------------------------------------------------------

for col in [
    "uk_region",
    "support_channel",
    "product_area",
    "priority_initial",
    "issue_category",
    "customer_segment"
]:
    clean[col] = clean[col].astype("string").str.strip()

# Standardise support channels
clean["support_channel"] = (
    clean["support_channel"]
    .str.lower()
    .replace({
        "e-mail": "email",
        "email": "email",
        "web": "web",
        "chat": "chat",
        "phone": "phone",
        "partner portal": "partner portal"
    })
)

# Standardise regions and product names
clean["uk_region"] = clean["uk_region"].str.title()
clean["product_area"] = clean["product_area"].str.title()

# ------------------------------------------------------------
# Convert dates
# ------------------------------------------------------------

clean["submitted_at"] = pd.to_datetime(
    clean["submitted_at"],
    errors="coerce"
)

clean["resolved_at"] = pd.to_datetime(
    clean["resolved_at"],
    errors="coerce"
)

# ------------------------------------------------------------
# Fix invalid values
# ------------------------------------------------------------

clean.loc[
    clean["agent_queue_length_at_submit"] < 0,
    "agent_queue_length_at_submit"
] = np.nan

# ------------------------------------------------------------
# Remove duplicate records
# ------------------------------------------------------------

clean = clean.drop_duplicates()

# ------------------------------------------------------------
# Handle missing numerical values
# ------------------------------------------------------------

clean["avg_sentiment_score"] = (
    clean["avg_sentiment_score"]
    .fillna(clean["avg_sentiment_score"].median())
)

clean["customer_tenure_months"] = (
    clean["customer_tenure_months"]
    .fillna(clean["customer_tenure_months"].median())
)

clean["agent_queue_length_at_submit"] = (
    clean["agent_queue_length_at_submit"]
    .fillna(clean["agent_queue_length_at_submit"].median())
)

# ------------------------------------------------------------
# Handle missing categorical values
# ------------------------------------------------------------

clean["product_area"] = (
    clean["product_area"]
    .fillna(clean["product_area"].mode()[0])
)

clean["priority_initial"] = (
    clean["priority_initial"]
    .fillna(clean["priority_initial"].mode()[0])
)

# ------------------------------------------------------------
# Validation Checks
# ------------------------------------------------------------

print("Shape after cleaning:", clean.shape)

print("\nRemaining Missing Values:")
display(
    clean.isna()
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

print("\nDuplicate Rows Remaining:")
print(clean.duplicated().sum())

print("\nNegative Queue Values Remaining:")
print(
    (clean["agent_queue_length_at_submit"] < 0).sum()
)
```

## Cleaning Results

The data-cleaning process successfully addressed the key quality issues identified during the initial assessment. Duplicate records were removed, reducing the dataset size from 15,090 to 15,000 observations. Inconsistent categorical labels were standardized to improve consistency across region, support channel, and product area variables.

Invalid queue length values were corrected and no negative queue values remained after cleaning. Missing numerical values were imputed using median values, while missing categorical values were replaced using the most frequently occurring category.

A small number of records (less than 1% of the dataset) contained missing submission timestamps. Because submission time is important for later feature engineering activities, these records will be removed to ensure data completeness.

Following cleaning, the dataset is considered suitable for exploratory analysis, feature engineering, and predictive modelling.
