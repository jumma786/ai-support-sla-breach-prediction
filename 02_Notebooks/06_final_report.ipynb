# AI-Driven SLA Breach Prediction for Customer Support Operations

## 1. Business Problem

Customer support teams must resolve tickets within agreed Service Level Agreements (SLAs). SLA breaches can lead to customer dissatisfaction, reputational damage, and increased operational costs. The objective of this project was to develop a machine learning solution capable of predicting whether a support ticket is likely to breach its SLA at the time of submission. Early identification of high-risk tickets enables proactive intervention, improved resource allocation, and better customer outcomes.

## 2. Data-Quality Findings

The dataset contained 15,090 records and 30 variables describing customer, ticket, operational, and agent-related characteristics. Several data-quality issues were identified during initial assessment. Missing values were present in sentiment scores, customer tenure, product area, priority level, and submission timestamps. Ninety duplicate records were detected and removed. Inconsistent category labels were observed across multiple fields, including support channels, product areas, and geographic regions. Additionally, 64 negative queue-length values were identified and treated as invalid observations. After cleaning, the final dataset contained 15,000 high-quality records suitable for analysis and modelling.

## 3. EDA Findings

Exploratory Data Analysis revealed that approximately 61% of tickets breached their SLA targets. Enterprise and Public Sector customers experienced significantly higher breach rates than SMB customers. Service outage and integration failure issues showed the highest probability of SLA breach, while access requests and how-to enquiries demonstrated substantially lower risk. Ticket priority was strongly associated with breach outcomes, with Critical tickets exhibiting extremely high breach rates. Operational factors such as queue length, backlog age, and response delays also showed clear relationships with SLA performance. Temporal analysis suggested that tickets submitted during peak business hours were more likely to breach SLA targets.

## 4. Feature Engineering Decisions

Feature engineering focused exclusively on information available at ticket submission time. Temporal features were derived from submission timestamps, including submission hour, day of week, month, and weekend indicators. Operational workload measures such as queue pressure and high backlog flags were created to capture support-team capacity constraints. Additional sentiment and text-based indicators were engineered from customer messages. Logarithmic transformation was applied to contract value to reduce skewness. Leakage variables including resolved_at, resolution_hours, final_priority, and escalated_prior_to_resolution were excluded from model training to ensure realistic deployment performance.

## 5. Model Comparison

Multiple classification models were evaluated using a consistent preprocessing pipeline and train-test split. Models included Dummy Classifier, Logistic Regression, Decision Tree, Random Forest, Gradient Boosting, and K-Nearest Neighbours. Gradient Boosting achieved the strongest overall performance with an F1-score of 0.903 and ROC-AUC of 0.961. Logistic Regression and Random Forest also demonstrated strong predictive capability, achieving F1-scores close to 0.90. All machine learning models substantially outperformed the baseline classifier, confirming the predictive value of the engineered features.

## 6. Final Evaluation

A Random Forest classifier was further evaluated using five-fold stratified cross-validation and hyperparameter optimisation. Cross-validation produced an average F1-score of 0.898 and ROC-AUC of 0.959, with low variability across folds, indicating stable and reliable performance. Threshold analysis demonstrated that a probability threshold of 0.50 provided the strongest balance between precision and recall. At this threshold, the model correctly identified the majority of SLA breaches while maintaining a manageable number of false positives. The final model achieved approximately 88% overall accuracy and successfully detected more than 90% of breached tickets.

## 7. Risk and Governance

Several risks should be considered before deployment. Changes in support processes, staffing levels, or customer behaviour may reduce model performance over time and require ongoing monitoring. Certain features, such as assigned agent information, may introduce operational bias if organisational structures change. Regular retraining and performance validation should therefore be implemented. Data governance procedures should ensure that customer information is handled securely and that model decisions remain transparent and explainable to business stakeholders.

## 8. Final Recommendation

The project successfully demonstrates that machine learning can accurately identify support tickets at risk of SLA breach before resolution activities begin. The recommended solution is the optimised Random Forest pipeline operating with a probability threshold of 0.50. This approach provides a strong balance between breach detection and operational efficiency, enabling proactive ticket management, improved resource allocation, and enhanced customer service performance. Future improvements could include real-time deployment, automated escalation workflows, and advanced resolution-time forecasting capabilities.
